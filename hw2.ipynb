#作業2：多隱藏層的全連接神經網路可批次學習

```python
import numpy as np
```

自行產生五個x各自的6個特徵值及其真實結果y

```python
X = np.array([
    [0, 2, 4, 6, 8, 10],
    [1, 3, 5, 7, 9, 0],
    [3, 6, 5, 9, 0, 1],
    [5, 3, 0, 7, 9, 8],
    [2, 1, 6, 3, 8, 5]
])

y = np.array([
    [1, 0, 0],
    [0, 0, 1],
    [0, 1, 0],
    [1, 0, 0],
    [0, 1, 0]
])

print("X =", X)
print("X.shape =", X.shape)

print("y =", y)
print("y.shape =", y.shape)
```

第一層權重 W1 與偏值 b1

```python
W1 = np.array([
    [0.5, 1.0, 2.0, 0.4, 1.5],
    [0.2, 0.5, 2.4, 1.5, 1.6],
    [0.4, 0.5, 1.5, 1.6, 0.8],
    [0.5, 1.5, 2.6, 0.4, 0.6],
    [0.4, 0.6, 0.2, 1.8, 0.5],
    [0.6, 0.2, 0.8, 0.4, 1.0]
])
b1 = np.array([1.0, 3.0, 2.0, 7.0, 6.0])

print("W1 =", W1)
print("b1 =", b1)
print("X.shape =", X.shape)
print("W1.shape =", W1.shape)
print("b1.shape =", b1.shape)
```

定義 ReLU

```python
def relu(x):
    return np.maximum(0, x)
```

```python
Z1 = np.dot(X, W1) + b1
H1 = relu(Z1)

print("Z1 =", Z1)
print("Z1.shape =", Z1.shape)
print("H1 =", H1)
print("H1.shape =", H1.shape)
```

第二層權重 W2 與偏值 b2

```python
W2 = np.array([
    [0.5, 1.0, 2.0, 0.4],
    [0.2, 0.5, 2.4, 1.5],
    [0.4, 0.5, 1.5, 1.6],
    [0.5, 1.5, 2.6, 0.4],
    [0.4, 0.6, 0.2, 1.8],
])

b2 = np.array([1.0, 3.0, 2.0, 7.0])


print("W2 =", W2)
print("b2 =", b2)
print("H1.shape =", H1.shape)
print("W2.shape =", W2.shape)
print("b2.shape =", b2.shape)
```

```python
Z2 = np.dot(H1, W2) + b2
H2 = relu(Z2)

print("Z2 =", Z2)
print("Z2.shape =", Z2.shape)
print("H2 =", H2)
print("H2.shape =", H2.shape)
```

第三層權重 W3 與偏值 b3

```python
W3 = np.array([
    [0.5, 1.0, 2.0, 0.4],
    [0.2, 0.5, 2.4, 1.5],
    [0.4, 0.5, 1.5, 1.6],
    [0.5, 1.5, 2.6, 0.4],
])

b3 = np.array([1.0, 3.0, 2.0, 7.0])

print("W3 =", W3)
print("b3 =", b3)
print("H2.shape =", H2.shape)
print("W3.shape =", W3.shape)
print("b3.shape =", b3.shape)
```

定義 Sigmoid

```python
def sigmoid(x):
    return 1/(1+np.exp(-x))
```

```python
Z3 = np.dot(H2, W3) + b3
H3 = sigmoid(Z3)

print("Z3 =", Z3)
print("Z3.shape =", Z3.shape)
print("H3 =", H3)
print("H3.shape =", H3.shape)
```

第四層(輸出層) 權重 W4與偏值 b4，計算 y_hat

```python
W4 = np.array([
    [0.5, 1.0, 2.0],
    [0.2, 0.5, 2.4],
    [0.4, 0.5, 1.5],
    [0.5, 1.5, 2.6],
])

b4 = np.array([1.0, 3.0, 2.0])

print("W4 =", W4)
print("b4 =", b4)
print("H3.shape =", H3.shape)
print("W4.shape =", W4.shape)
print("b4.shape =", b4.shape)
```

定義Softmax

```python
def softmax(x):
    x = x - np.max(x, axis=-1, keepdims=True)
    return np.exp(x) / np.sum(np.exp(x), axis=-1, keepdims=True)
```

```python
Z4 = np.dot(H3, W4) + b4
y_hat = softmax(Z4)

print("Z4 =", Z4)
print("Z4.shape =", Z4.shape)
print("y_hat =", y_hat)
print("y_hat.shape =", y_hat.shape)
```

cross_entropy_error計算y_hat和y誤差值的平均

定義Cross entropy error (交叉熵誤差)

```python
def cross_entropy_error(y_hat, t):
    if y_hat.ndim == 1:
        t = t.reshape(1, t.size)
        y_hat = y_hat.reshape(1, y_hat.size)

    batch_size = y_hat.shape[0]

    return -np.sum(t * np.log(y_hat + 1e-7)) / batch_size
```

y_hat和y誤差值的平均

```python
cross_entropy_error(y_hat, y)
```
